```python
import time

def print_numbers():
    for i in range(10):
        print(f"Number: {i}")
        time.sleep(2)
def print_letters():
    for letter in "ABCDEFGHIJK":
        print(f"Letter: {letter}")
        time.sleep(2)
start_time = time.time()
print_numbers()
print_letters()
end_time = time.time()
print(f"Total time: f{end_time - start_time}")
```

```python
import time
import threading
def print_numbers():
    for i in range(10):
        print(f"Number: {i}")
        time.sleep(1)
def print_letters():
    for letter in "ABCDEFGHIJK":
        print(f"Letter: {letter}")
        time.sleep(1)
start_time = time.time()
t1 =threading.Thread(target=print_numbers)
t2 =threading.Thread(target=print_letters)

t1.start()
t2.start()
t1.join()
t2.join()
end_time = time.time()
print(f"Total time: f{end_time - start_time}")
```

```python
import time
import threading
def print_numbers():
    for i in range(5):
        print(f"Number: {i}")
        time.sleep(2)
def print_letters():
    for letter in "ABC":
        print(f"Letter: {letter}")
        time.sleep(3)
def print_symbol():
    for letter in "@!#%$":
        print(f"Letter: {letter}")
        time.sleep(1)
start_time = time.time()
t1 =threading.Thread(target=print_numbers)
t2 =threading.Thread(target=print_letters)
t3 =threading.Thread(target=print_symbol)

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()
end_time = time.time()
print(f"Total time: f{end_time - start_time}")
```
