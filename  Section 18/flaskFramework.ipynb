```python
from flask import Flask
app = Flask(__name__)

@app.route("/")
def home():
    return "<h1>This is a Home Page</h1>"

@app.route("/index") 
def index():
    return "This is a Index Page"

@app.route("/about")
def about():
    return "This is a About Page"
if __name__ == "__main__":
    app.run()
```
