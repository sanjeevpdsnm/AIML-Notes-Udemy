```python
class Vector:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    # addition
    def __add__(self, other):
        return Vector(self.x + other.x,
                      self.y + other.y)

    # subtraction
    def __sub__(self, other):
        return Vector(self.x - other.x,
                      self.y - other.y)

    # multiplication by scalar
    def __mul__(self, other):
        return Vector(self.x * other.x,
                      self.y * other.y)

    # equality check
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    # printable representation
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
v1 = Vector(2,3)
v2 = Vector(5,4)
```

```python
print(v1+v2)
print(v1*v2)
print(v1-v2)
print(v1 != v2)
```

```python
def square(n):
    for i in range(n):
        yield i*i
print(square(10))  
for i in square(10):
    print(i)
```

```python
def deco(func):
    def wrapper():
        print("before")
        func()
        print("after")
    return wrapper
```

```python
@deco
def hello():
    print("hello")
hello()
```
