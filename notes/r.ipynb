# 🔥 1. Reading JSON → DataFrame

---

## JSON input

```python
import pandas as pd
from io import StringIO

json_data = """
{
  "employee_name": "James",
  "email": "james@gmail.com",
  "job_profile": [
    {"title1": "Team Lead"},
    {"title2": "Senior Developer"}
  ]
}
"""
```

---

## Convert JSON → DataFrame

```python
df = pd.read_json(StringIO(json_data))
print(df)
```

### Output:

```
  employee_name            email            job_profile
0        James  james@gmail.com  {'title1': 'Team Lead'}
1        James  james@gmail.com  {'title2': 'Senior Developer'}
```

Pandas expands nested JSON into rows.

---

# 🔥 Convert DataFrame → JSON

```python
df.to_json()
```

Default orientation = index-based.

Output example:

```
{"employee_name":{"0":"James","1":"James"},
 "email":{"0":"james@gmail.com","1":"james@gmail.com"}}
```

---

## Use record format (clean JSON)

```python
df.to_json(orient="records")
```

Output:

```
[
 {"employee_name":"James","email":"james@gmail.com"},
 {"employee_name":"James","email":"james@gmail.com"}
]
```

This matches API-style JSON.

# 🔥 2. Reading CSV From URL

---

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data"

df = pd.read_csv(url, header=None)
df.head()
```

### Output:

```
0   1    2    3   ...
1 14.23 1.71 2.43 ...
1 13.20 1.78 2.14 ...
...
```

Header=None → pandas auto-index columns.

---

## Save back to CSV

```python
df.to_csv("wine.csv", index=False)
```

Now file exists locally.

# 🔥 3. Reading HTML Tables

Pandas can scrape tables directly.

---

## Install dependencies

```bash
pip install lxml html5lib beautifulsoup4
```

These are required.

---

## Read HTML

```python
url = "https://en.wikipedia.org/wiki/Mobile_country_code"

tables = pd.read_html(url)
```

`tables` is a list of DataFrames.

---

## Access first table

```python
df = tables[0]
df.head()
```

Output:

```
Country    MCC    MNC
India      404    ...
USA        310    ...
...
```

Automatic web scraping.

---

## Match specific table

```python
tables = pd.read_html(url, match="Country")
df = tables[0]
```

Only tables containing keyword “Country” are returned.

Professional scraping technique.

# 🔥 5. Convert to Pickle

Pickle = serialized Python object.

Fast load/save for ML pipelines.

---

## Save DataFrame

```python
df.to_pickle("data.pkl")
```

File becomes binary serialized.

---

## Load pickle

```python
df = pd.read_pickle("data.pkl")
```

Restores DataFrame instantly.

No parsing needed → extremely fast.

Used heavily in ML production systems.
