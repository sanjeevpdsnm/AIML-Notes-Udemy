```python
import os

structure = {
    "github-portfolio-analyzer": {
        "backend": {
            "app": {
                "__init__.py": "",
                "main.py": "",
                "api": {
                    "__init__.py": "",
                    "routes.py": "",
                    "models.py": "",
                },
                "core": {
                    "__init__.py": "",
                    "github_client.py": "",
                    "analyzer.py": "",
                    "scorer.py": "",
                    "recommendations.py": "",
                },
                "utils": {
                    "__init__.py": "",
                    "cache.py": "",
                    "helpers.py": "",
                },
            },
            "requirements.txt": "",
            ".env.example": "",
            "Dockerfile": "",
        },
        "frontend": {
            "public": {
                "index.html": "",
            },
            "src": {
                "App.js": "",
                "index.js": "",
                "components": {
                    "ProfileInput.js": "",
                    "LoadingSpinner.js": "",
                    "ScoreCard.js": "",
                    "SignalsList.js": "",
                    "Recommendations.js": "",
                    "RepoList.js": "",
                    "ReportPage.js": "",
                },
                "styles": {
                    "tailwind.css": "",
                },
            },
            "package.json": "",
            "tailwind.config.js": "",
        },
        "docker-compose.yml": "",
        "README.md": "",
        ".gitignore": "",
    }
}


def create_structure(base_path, tree):
    for name, content in tree.items():
        path = os.path.join(base_path, name)
        if isinstance(content, dict):
            os.makedirs(path, exist_ok=True)
            create_structure(path, content)
        else:
            with open(path, "w") as f:
                f.write(content)


if __name__ == "__main__":
    create_structure(".", structure)
    print("✅ Project folder structure created successfully!")
```
