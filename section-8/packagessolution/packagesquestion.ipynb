```python
import os
import zipfile

structure = [
    "app/__init__.py",
    "app/main.py",
    "app/config.py",
    "app/database.py",

    "app/auth/__init__.py",
    "app/auth/models.py",
    "app/auth/schemas.py",
    "app/auth/services.py",
    "app/auth/dependencies.py",
    "app/auth/routes.py",

    "app/urls/__init__.py",
    "app/urls/models.py",
    "app/urls/schemas.py",
    "app/urls/services.py",
    "app/urls/dependencies.py",
    "app/urls/routes.py",

    "app/analytics/__init__.py",
    "app/analytics/models.py",
    "app/analytics/schemas.py",
    "app/analytics/services.py",
    "app/analytics/routes.py",

    "app/common/__init__.py",
    "app/common/exceptions.py",
    "app/common/logging.py",
    "app/common/cache.py",

    "app/tests/__init__.py",
    "app/tests/conftest.py",
    "app/tests/test_auth.py",
    "app/tests/test_urls.py",
    "app/tests/test_analytics.py",

    "alembic/versions/",
    ".env.example",
    "requirements.txt",
    "docker-compose.yml",
    "README.md",
]

root = "url-shortener"

for path in structure:
    full = os.path.join(root, path)
    os.makedirs(os.path.dirname(full), exist_ok=True)
    if not path.endswith("/"):
        open(full, "w").close()

zip_name = "url-shortener.zip"
with zipfile.ZipFile(zip_name, "w", zipfile.ZIP_DEFLATED) as z:
    for foldername, _, filenames in os.walk(root):
        for f in filenames:
            file_path = os.path.join(foldername, f)
            z.write(file_path)

print(f"✅ Zip created: {zip_name}")
```
