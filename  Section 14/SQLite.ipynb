```python
import sqlite3

connection = sqlite3.connect("example.db")
cursor = connection.cursor()
```

```python
cursor.execute("""
               Create table if not exists employees (
                   id integer primary key,
                   name Char,
                   age Integer,
                   department char
               )
               """)

connection.commit()
```

```python
cursor.execute("""
               select * from  employees
               
               """)
```

```python
cursor.execute(
    """
    Insert into employees (name,age,department) 
    values ("Sanjeev",21,"Data Scientist")
    """
)
```

```python
cursor.execute("Select * from employees")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute("""
               Update employees 
               set department = "Entrepenuer"
               where name = "Sanjeev"
               """)
connection.commit()
```

```python
cursor.execute("Select * from employees")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute("""
               Delete from employees
               where name = "Sanjeev"
               """)
connection.commit()
```
