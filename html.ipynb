# 🌐 1. Introduction to HTML

### 📌 Definition (VERY IMPORTANT for exams)

**HTML (Hyper Text Markup Language)** is the standard language used to **create and structure web pages and their content**.

👉 As shown on **page 3**, HTML is used for:

* Structure/layout of webpage
* Content arrangement

### 🧠 Key Points:

* HTML is **not a programming language**
* It is a **markup language**
* Uses **tags** to define elements

---

# 🧱 2. HTML Tags & Elements

### 📌 Definition

An **HTML tag** is a container used to hold content.

👉 Example (page 5):

```html
<p>This is a paragraph</p>
```

### 🔍 Explanation:

* `<p>` → opening tag
* `</p>` → closing tag
* Content → "This is a paragraph"

### 📌 Exam Point:

👉 Combination of **tag + content = element**

---

# 🏗️ 3. Basic HTML Structure (MOST IMPORTANT)

👉 Given on **page 6**

```html
<!DOCTYPE html>
<html>
<head>
<title>My First Page</title>
</head>
<body>
<p>Hello World</p>
</body>
</html>
```

### 🔍 Explanation:

| Tag               | Meaning         |
| ----------------- | --------------- |
| `<!DOCTYPE html>` | Defines HTML5   |
| `<html>`          | Root element    |
| `<head>`          | Metadata        |
| `<title>`         | Page title      |
| `<body>`          | Visible content |

### 📌 Exam Tip:

👉 Always draw **structure diagram** for 5–8 marks.

---

# 🎨 1. Introduction to CSS

### 📌 Definition (VERY IMPORTANT)

**CSS (Cascading Style Sheets)** is a language used to **describe the style and appearance of HTML documents**.

👉 As shown on **page 3**:

* CSS is a **styling language (not programming language)**
* It works **with HTML content**

---

# 🧩 2. Basic CSS Syntax

👉 Page 4

```css
h1 {
  color: red;
}
```

### 🔍 Explanation:

* `h1` → Selector
* `color` → Property
* `red` → Value

### 📌 Exam Point:

👉 CSS syntax = **Selector + Property + Value**

---

# 🧷 3. Ways to Add CSS (IMPORTANT)

👉 Page 5–6

## 🔹 1. Inline CSS

```html
<h1 style="color:red">Hello</h1>
```

## 🔹 2. Internal CSS

```html
<style>
h1 {
  color: red;
}
</style>
```

## 🔹 3. External CSS (BEST)

```html
<link rel="stylesheet" href="style.css">
```

### 📌 Priority Order:

👉 Inline > Internal > External

---

## ✅ Definition (Very Important ⭐)

**The Internet is a global network of interconnected computers and devices that communicate using standardized protocols like TCP/IP.** 

---

## 🧠 Explanation (5–10 Marks)

* Internet connects millions of devices worldwide
* Uses **packet-based communication**:

  * Data is broken into packets
  * Packets travel via routers
  * Reassembled at destination

👉 From Page 1:

* Global connectivity
* Decentralized system
* Uses TCP/IP

---

* Packet switching
