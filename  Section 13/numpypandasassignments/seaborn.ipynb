```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.scatterplot(
    x ="total_bill",
    y = "tip",
    data = tips
)

plt.title("Total Bill vs Tip")
plt.show()
```

```python
sns.lineplot(
    x ="total_bill",
    y = "tip",
    data = tips
)

plt.title("Total Bill vs Tip")
plt.show()
```

```python
sns.barplot(
    x ="total_bill",
    y = "tip",
    data = tips
)

plt.title("Total Bill vs Tip")
plt.show()
```

```python
import numpy as np
sns.barplot(
    x ="total_bill",
    y = "tip",
    data = tips,
    estimator = np.sum
)
plt.show()
```

```python
sns.violinplot(
    x ="total_bill",
    y = "tip",
    data = tips
)

plt.title("Total Bill vs Tip")
plt.show()
```

```python
sns.histplot(
    tips["total_bill"],
    bins = 10,
    kde= True
)

plt.title("Total Bill vs Tip")
plt.show()
```

```python
sns.kdeplot(
    tips["total_bill"],
    fill = True
)
```

```python
sns.pairplot(tips)
plt.show()
```

```python
corr = tips[["total_bill", "tip", "size"]].corr()
corr
```

```python
sns.heatmap(
    corr,
    annot=True,
    cmap = "coolwarm"
) 
plt.show()
```

```python
sns.set_style("darkgrid")
sns.set_palette("Set2")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="sex",
    data=tips
)

plt.show()
```

```python
sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    size="size",
    data=tips
)
plt.show()
```

```python
sns.jointplot(
    x="total_bill",
    y="tip",
    data=tips,
    kind="scatter"
)
plt.show()
```

```python
sns.jointplot(
    x="total_bill",
    y="tip",
    data=tips,
    kind="scatter"
)
```

```python

sns.jointplot(
    x="total_bill",
    y="tip",
    data=tips,
    kind="reg"
)
```

```python
sns.regplot(
    x="total_bill",
    y="tip",
    data=tips
)
plt.show()
```

```python
g = sns.FacetGrid(tips,col="time")
g.map(sns.scatterplot,"total_bill","tip")
plt.show()
```

```python
g =sns.FacetGrid(tips,row="sex",col="time")
g.map(sns.scatterplot,"total_bill","tip")
plt.show()
```

```python
sns.catplot(
    x="day",
    y="total_bill",
    hue="sex",
    kind="bar",
    data=tips
)
```

```python
sns.catplot(
    x="day",
    y="total_bill",
    kind="box",
    data=tips
)
```

```python
sns.catplot(
    x="day",
    y="total_bill",
    kind="violin",
    data=tips
)
```

```python
corr = tips[["total_bill","tip","size"]].corr()
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=1
)
```

```python
import numpy as np
mask = np.triu(np.ones_like(corr))

sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    cmap="coolwarm"
)
sns.set_theme(style="dark")
```
