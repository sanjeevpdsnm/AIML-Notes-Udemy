```python
import gc
gc.enable()
gc.disable()
print(gc.collect())
# print(gc.get_stats())

for d in gc.get_stats() :
    print(d)
```

```python
class MyObject:
    def __init__(self,name):
        self.name = name
        print(f"Object {self.name} created")
    def __del__(self):
        print(f"Object {self.name} deleted")
obj1 = MyObject("obj1")
obj2 = MyObject("obj2")

obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2

gc.collect()
```

```python
def generate_numbers(n):
    for i in range(n):
        yield i
        
for num in generate_numbers(10000000000):
    if num > 10:
        break
    print(num)
```

```python
import tracemalloc
tracemalloc.start()

def create_list():
    return [i for i in range(10000)]
create_list()

snapshot = tracemalloc.take_snapshot()
top_stats = snapshot.statistics("lineno")

for stat in top_stats[:]:
    print(stat)
```
