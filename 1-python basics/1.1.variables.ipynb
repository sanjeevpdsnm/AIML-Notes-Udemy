```python
num1 = int(input('Enter first number: '))
num2 = int(input('Enter second number: '))
sum_ = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2
print('Sum:', sum_)
print('Difference:', difference)
print('Product:', product)
print('Quotient:', quotient)
print(type(sum_))
print(type(difference))
print(type(product))
print(type(quotient))
```
