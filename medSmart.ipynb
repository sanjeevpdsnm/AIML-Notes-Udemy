# 📄 SOFTWARE REQUIREMENTS SPECIFICATION (SRS)

## **Project: MedSmart – Medicine Comparison & Redirection Platform**

---

# 1. 🧭 INTRODUCTION

## 1.1 Purpose

This document defines the requirements for **MedSmart**, a web-based platform that allows users to search medicines, compare prices across multiple online pharmacy platforms, and redirect users to those platforms for purchase.

The system is designed as an **aggregator**, not a direct seller, to avoid inventory, logistics, and regulatory complexities.

---

## 1.2 Scope

MedSmart will:

* Allow users to search for medicines
* Display medicine details and price comparisons
* Redirect users to third-party platforms (affiliate model)
* Track user interactions for analytics

MedSmart will NOT:

* Sell medicines directly
* Handle delivery or logistics
* Store or manage inventory

---

## 1.3 Intended Audience

* Developers (Frontend/Backend Engineers)
* Project Evaluators (Faculty)
* System Designers
* Startup/Product Teams

---

## 1.4 Definitions

| Term           | Description                                                    |
| -------------- | -------------------------------------------------------------- |
| Aggregator     | Platform that collects and displays data from multiple sources |
| Affiliate Link | Trackable URL used to earn commission                          |
| API            | Interface to communicate with external systems                 |
| Redirect       | Sending user to another website                                |

---

# 2. 🧠 OVERALL DESCRIPTION

## 2.1 Product Perspective

MedSmart is a **standalone web application** that integrates with external pharmacy platforms like Tata 1mg and PharmEasy.

It follows a **3-tier architecture**:

* Frontend (UI)
* Backend (Business Logic)
* Database (Data Storage)

---

## 2.2 Product Functions

* Medicine search
* Price comparison
* Affiliate redirection
* User activity tracking
* Admin analytics

---

## 2.3 User Classes

| User            | Description                      |
| --------------- | -------------------------------- |
| Guest           | Search and view medicines        |
| Registered User | Optional login, preferences      |
| Admin           | Manage system data and analytics |

---

## 2.4 Operating Environment

* Browser: Chrome, Edge, Firefox
* Frontend: React.js
* Backend: Node.js (Express)
* Database: MongoDB
* Hosting: Cloud (AWS / Vercel / Render)

---

## 2.5 Constraints

* No direct selling due to legal restrictions
* Dependence on third-party platforms
* Limited API access from pharmacy platforms
